Plan is to pull in some Jordy Nelson game-level data and then do some simple bayesian analyses.

```python
import nflreadpy as nfl # type: ignore
import polars as pl
```

```python
# pull jordy details
jordy = nfl.load_players().filter((pl.col("first_name") == "Jordy") & (pl.col("last_name") == "Nelson"))
```

```python
# download jordy's stats by season
first_season = jordy["rookie_season"].first()
last_season = jordy["last_season"].first()
jordy_id = jordy["gsis_id"].first()
season_range = list(range(first_season, last_season+1))
```

```python
id_predicate = pl.col("player_id") == jordy_id
jordy_stats = nfl.load_player_stats(season_range).filter(id_predicate)
```
